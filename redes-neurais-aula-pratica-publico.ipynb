```python
import numpy as np
import matplotlib.pyplot as plt
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Dense
from tensorflow.keras.optimizers import Adam
```

```python
# Geração de dados sintéticos
np.random.seed(42)
X = np.linspace(0, 10, 100)
y = 2 * X + np.random.randn(100) * 2

plt.scatter(X, y, color='blue')
plt.title("Dados para Regressão Linear com Ruído")
plt.xlabel("X")
plt.ylabel("y")
plt.grid(True)
plt.show()
```

```python
modelo = Sequential([
    Dense(4, activation='relu', input_shape=(1,)),
    Dense(1)
])

modelo.compile(optimizer=Adam(learning_rate=0.01), loss='mse')

hist1 = modelo.fit(X, y, epochs=200, verbose=0)
```

```python
modelo2 = Sequential([
    Dense(10, activation='relu', input_shape=(1,)),
    Dense(1)
])

modelo2.compile(optimizer=Adam(learning_rate=0.01), loss='mse')

hist2 = modelo2.fit(X, y, epochs=200, verbose=0)
```

```python
modelo3 = Sequential([
    Dense(4, activation='relu', input_shape=(1,)),
    Dense(1)
])

modelo3.compile(optimizer=Adam(learning_rate=0.1), loss='mse')

hist3 = modelo3.fit(X, y, epochs=200, verbose=0)
```

```python
plt.figure(figsize=(8,5))
plt.plot(hist1.history['loss'], label='4 neurônios, lr=0.01')
plt.plot(hist2.history['loss'], label='10 neurônios, lr=0.01')
plt.plot(hist3.history['loss'], label='4 neurônios, lr=0.1')

plt.title("Comparação de Hiperparâmetros")
plt.xlabel("Épocas")
plt.ylabel("Erro (MSE)")
plt.legend()
plt.grid(True)
plt.show()
```

```python
y_pred = modelo2.predict(X)

plt.scatter(X, y, label='Dados reais')
plt.plot(X, y_pred, color='red', label='Previsão do modelo')
plt.title("Ajuste do Modelo aos Dados")
plt.xlabel("X")
plt.ylabel("y previsto")
plt.legend()
plt.show()
```

```python
y_pred1 = modelo.predict(X)
y_pred2 = modelo2.predict(X)
y_pred3 = modelo3.predict(X)

plt.scatter(X, y, label='Dados reais')

plt.plot(X, y_pred1, label='Modelo 1 (4 neurônios, lr=0.01)')
plt.plot(X, y_pred2, label='Modelo 2 (10 neurônios, lr=0.01)')
plt.plot(X, y_pred3, label='Modelo 3 (4 neurônios, lr=0.1)')

plt.title("Comparação dos Modelos")
plt.xlabel("X")
plt.ylabel("y previsto")
plt.legend()
plt.show()
```
